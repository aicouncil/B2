```python
no = '/kaggle/input/brain-mri-images-for-brain-tumor-detection/no'
yes = '/kaggle/input/brain-mri-images-for-brain-tumor-detection/yes'
```

```python
import os

features = []

for img_name in os.listdir(no):
    #print(os.path.join(no, img_name))
    img_loc = os.path.join(no, img_name)
    image = cv2.imread(img_loc)
    image = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)
    image = cv2.resize(image , (100,100))
    features.append(image)

for img_name in os.listdir(yes):
    #print(os.path.join(yes, img_name))
    img_loc = os.path.join(yes, img_name)
    image = cv2.imread(img_loc)
    image = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)
    image = cv2.resize(image , (100,100))
    features.append(image)
```

```python
import numpy as np
features = np.array(features)
features.shape
```

```python
labels = []

for img_name in os.listdir(no):
    #print(os.path.join(no, img_name))
    labels.append(0)


for img_name in os.listdir(yes):
    #print(os.path.join(yes, img_name))
    labels.append(1)

labels = np.array(labels)
labels.shape
```

```python
X = features
y = labels
```
