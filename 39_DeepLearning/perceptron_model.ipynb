```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/bipulshahi/Dataset/refs/heads/main/Advertising.csv',
                 index_col=0)
df.head()
```

```python
X = df[['TV', 'radio' ,	'newspaper']]
y = df['sales']
```

```python
#Yh = w0 + w1X1 + w2X2 + w3X3

w0 = 1
w1 = 1
w2 = 1
w3 = 1

Yh = w0 + w1*X.TV + w2*X.radio + w3*X.newspaper          #feed forward process
print(Yh)
```

```python
error = ((y - Yh)**2).mean()
print(error)
```

```python
#Cacluate gradient

w0 = 1
w1 = 1
w2 = 1
w3 = 1

for i in range(0,1000):
  Yh = w0 + w1*X.TV + w2*X.radio + w3*X.newspaper

  dew0 = -2*((y - Yh)).mean()
  dew1 = -2*((y - Yh)*X.TV).mean()
  dew2 = -2*((y - Yh)*X.radio).mean()
  dew3 = -2*((y - Yh)*X.newspaper).mean()

  lr = 1e-5

  w0 = w0 - lr * dew0
  w1 = w1 - lr * dew1
  w2 = w2 - lr * dew2
  w3 = w3 - lr * dew3

  error = ((y - Yh)**2).mean()
  print(i,error)
```

```python
print(w0 , w1 , w2 , w3)
```

```python
mae = abs(y - Yh).mean()
print(mae)
```

**Neural network using keras**

```python
#Step 1 - Data definition
X = df[['TV', 'radio' ,	'newspaper']]
y = df[['sales']]
```

```python
#Step 2 - build the nueral network model
from keras import models,layers

model = models.Sequential()

model.add(layers.Dense(1 , input_shape = (3,)))
```

```python
model.summary()
```

```python
#Step 3 - compile the neural network model
from keras import optimizers
sgd = optimizers.SGD(learning_rate=1e-5)

model.compile(loss = 'mse', optimizer=sgd , metrics=['mae'])
```

```python
#Step 4 - Train your model
model.fit(X,y , epochs= 250)
```

```python
Yp = model.predict(X)
```

```python
abs(y - Yp).mean()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,y)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

scaler.fit(xtrain)

xtrainScaled = scaler.transform(xtrain)
xtestScaled = scaler.transform(xtest)
```

```python
xtrainScaled.shape
```

```python
#Step 2 - build the nueral network model
from keras import models,layers

modelB = models.Sequential()

modelB.add(layers.Dense(2 , activation='relu' , input_shape = (3,)))

modelB.add(layers.Dense(1 , activation='relu'))
```

```python
modelB.summary()
```

```python
sgd = optimizers.SGD(learning_rate=1e-2)
modelB.compile(loss = 'mse', optimizer=sgd , metrics=['mae'])
```

```python
modelB.fit(xtrainScaled,ytrain , epochs=100, validation_data=(xtestScaled,ytest))
```

```python
modelB.evaluate(xtrainScaled,ytrain)
```

```python
modelB.evaluate(xtestScaled,ytest)
```
