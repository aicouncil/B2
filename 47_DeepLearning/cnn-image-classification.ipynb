```python
cats = '/kaggle/input/dog-and-cat-classification-dataset/PetImages/Cat'
dogs = '/kaggle/input/dog-and-cat-classification-dataset/PetImages/Dog'
```

**Read all the images and store into an array - features**

```python
import cv2
import os
from tqdm import tqdm

features = []
labels = []

for img_name in tqdm(os.listdir(cats)):
    #print(os.path.join(cats, img_name))
    img = cv2.imread(os.path.join(cats, img_name))

    #Skip if img not loaded
    if img is None:
        print(f"Skipping file: {img_name}" )
        continue

    labels.append(0)
        
    img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img = cv2.resize(img, (50,50))
    features.append(img)


for img_name in tqdm(os.listdir(dogs)):
    #print(os.path.join(cats, img_name))
    img = cv2.imread(os.path.join(dogs, img_name))

    #Skip if img not loaded
    if img is None:
        print(f"Skipping file: {img_name}" )
        continue

    labels.append(1)
        
    img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img = cv2.resize(img, (50,50))
    features.append(img)
```

```python
import numpy as np
X = np.array(features)
y = np.array(labels)
```

```python
X = X/X.max()
```

```python
X.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(X,y)
```

```python
import matplotlib.pyplot as plt
plt.imshow(X[10000], cmap='gray')
plt.show()
```

```python
from keras import models,layers

model = models.Sequential()

model.add(layers.Conv2D(32 , (3,3) , padding='same', input_shape=(50,50,1)))
model.add(layers.MaxPooling2D(2,2))

model.add(layers.Conv2D(64 , (3,3) , padding='same'))
model.add(layers.MaxPooling2D(2,2))

model.add(layers.Conv2D(128 , (3,3) , padding='same'))
model.add(layers.MaxPooling2D(2,2))

model.add(layers.Conv2D(256 , (3,3) , padding='same'))
model.add(layers.MaxPooling2D(2,2))

model.add(layers.Conv2D(512 , (3,3) , padding='same'))
model.add(layers.MaxPooling2D(2,2))

model.add(layers.Flatten())

model.add(layers.Dense(10))

model.add(layers.Dense(1, activation = 'sigmoid'))
```

```python
model.summary()
```

```python
model.compile(loss = 'binary_crossentropy',
             optimizer = 'sgd',
             metrics = ['accuracy'])
```

```python
model.fit(xtrain, ytrain,
         validation_data = (xtest,ytest),
         epochs=20)
```

```python
model.evaluate(xtrain,ytrain)
```

```python
model.evaluate(xtest,ytest)
```
