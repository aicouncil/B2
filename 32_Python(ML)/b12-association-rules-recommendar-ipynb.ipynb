```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/kaggle/input/online-retail-ii-uci/online_retail_II.csv')
df.head(20)
```

```python
df['Invoice'].nunique()    #basket
```

```python
df['Customer ID'].nunique()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df1 = df.dropna()
df1.head()
```

```python
df1.shape
```

```python
df1['Customer ID'].nunique()
```

```python
df1['Description'].nunique()
```

```python
df1['Quantity'].describe()
```

```python
#Remove the instances from the data where product got cancelled
df2 = df1[~(df1['Quantity'] < 1)]
df2.head()
```

```python
df2.duplicated().sum()
```

```python
#Remove duplicates
df3 = df2.drop_duplicates()
df3.head()
```

```python
#Convert all characters into small letters
df3['Description'] = df3['Description'].str.lower()

#Remove any leading or lagging spaces in text of description column
df3['Description'] = df3['Description'].str.strip()
```

```python
df3.groupby(['Customer ID','Description'])['Quantity'].sum().unstack()
```

```python
df_basket = df3.groupby(['Customer ID','Description'])['Quantity'].sum().unstack().fillna(0)
df_basket
```

```python
df_basket.columns
```

```python
#All non zero cells into True and cells with value 0 into False

df_basket = df_basket.apply(lambda x : x > 0)
```

```python
df_basket.head()
```

```python
df_basket.shape
```

```python
#Identify most frequent itemsets
from mlxtend.frequent_patterns import apriori
frequent_items = apriori(df_basket, use_colnames=True, min_support=0.06)
frequent_items
```

```python
#Apply apripori algorithm to build recommendation system

from mlxtend.frequent_patterns import association_rules
rules = association_rules(frequent_items, metric='lift' , min_threshold=1)
rules
```

```python
r1 = rules[['antecedents', 'consequents', 'support','confidence', 'lift']]
r1
```

```python
r1['antecedents'] = r1['antecedents'].apply(lambda x : list(x))
r1['consequents'] = r1['consequents'].apply(lambda x : list(x))
r1.head()
```

```python
r1.to_csv('recommendar.csv')
```

```python
df_uk = df_uk.dropna(subset = ['Description'])
df_uk.head()
```

```python
df_uk['Description'] = df_uk['Description'].str.lower()
df_uk['Description'] = df_uk['Description'].str.strip()
```

```python
df_uk.head()
```

```python
df_uk = df_uk.drop_duplicates()
```

```python
df_uk.shape
```

```python
df_uk.head()
```

```python
df_uk['Quantity'].describe()
```

```python
df_uk = df_uk[~(df_uk['Quantity'] < 1)]
df_uk.head()
```

```python
df_uk['Invoice'].nunique()
```

```python
df_uk['Description'].nunique()
```

```python
#df_uk.groupby(['Invoice' , 'Description'])['Quantity'].sum().unstack()
```

```python
df_uk_basket = pd.crosstab(df_uk['Invoice'],df_uk['Description'])
df_uk_basket.head()
```

```python
df_uk_basket = df_uk_basket.apply(lambda x : x > 0)
df_uk_basket.head()
```

```python
#Most frequent itemsets
from mlxtend.frequent_patterns import apriori
frequent_items = apriori(df_uk_basket, use_colnames=True, min_support=0.01)
frequent_items
```

```python
#apply apriori to create recommendar table
from mlxtend.frequent_patterns import association_rules
res = association_rules(frequent_items,metric='lift',min_threshold=1)
res
```

```python
r2 = res[['antecedents', 'consequents', 'support','confidence', 'lift']]
r2
```

```python
r2['antecedents'] = r2['antecedents'].apply(lambda x : list(x))
r2['consequents'] = r2['consequents'].apply(lambda x : list(x))
r2.head()
```

```python
r2.to_csv('invoice_recommendar.csv')
```
