```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('C:/Users/vipul/Downloads/mnist_digits_train.csv')
df.head()
```

```python
df['784'].unique()
```

**Step 1** - Define the data

```python
X = df.drop(columns = ['784'])
y = df['784']
```

```python
X.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

Xs = scaler.fit_transform(X)
```

```python
#plt.imshow(X.values[90].reshape(28,28) , cmap='gray')
#plt.show()
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(Xs,y)
```

```python
xtrain.shape
```

**Step 2** - Define neural network model

```python
from keras import models, layers

model = models.Sequential()

model.add(layers.Dense(10 , input_shape = (784,)))

model.add(layers.Dense(1))
```

```python
model.summary()
```

**Step 3** - Compile the model

```python
from keras import optimizers
sgd = optimizers.SGD(learning_rate = 0.01)

model.compile(optimizer = sgd, loss = 'mse', metrics = ['accuracy'])
```

**Step 4**- Train the model

```python
model.fit(xtrain,ytrain,
         epochs = 10,
         validation_data = (xtest,ytest))
```
