```python
no = '/kaggle/input/brain-mri-images-for-brain-tumor-detection/no'
yes = '/kaggle/input/brain-mri-images-for-brain-tumor-detection/yes'
```

```python
import os

features = []

for img_name in os.listdir(no):
    #print(os.path.join(no, img_name))
    img_loc = os.path.join(no, img_name)
    image = cv2.imread(img_loc)
    image = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)
    image = cv2.resize(image , (100,100))
    features.append(image)

for img_name in os.listdir(yes):
    #print(os.path.join(yes, img_name))
    img_loc = os.path.join(yes, img_name)
    image = cv2.imread(img_loc)
    image = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)
    image = cv2.resize(image , (100,100))
    features.append(image)
```

```python
import numpy as np
features = np.array(features)
features.shape
```

```python
labels = []

for img_name in os.listdir(no):
    #print(os.path.join(no, img_name))
    labels.append(0)


for img_name in os.listdir(yes):
    #print(os.path.join(yes, img_name))
    labels.append(1)

labels = np.array(labels)
labels.shape
```

```python
X = features
y = labels
```

```python
X.shape
```

```python
import pandas as pd
df = pd.DataFrame(X.reshape(253,10000))
print(df.shape)
df.head()
```

```python
df.insert(0 , "Labels" , y)
df.head()
```

```python
Xt = X.reshape(253,10000)
```

```python
plt.imshow(Xt[55].reshape(100,100) , cmap='gray')
plt.show()
```

```python
Xt = Xt/Xt.max()
```

```python
Xt
```

```python
plt.imshow(Xt[55].reshape(100,100) , cmap='gray')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(Xt,y)
```

```python
xtrain.shape
```

```python
from keras import models, layers
```

```python
model = models.Sequential()

model.add(layers.Dense(128, activation = 'relu' , input_shape=(10000,)))

model.add(layers.Dense(64, activation = 'relu'))

model.add(layers.Dense(1, activation = 'sigmoid'))
```

```python
model.summary()
```

```python
model.compile(loss = 'binary_crossentropy',
             optimizer = 'adam',
             metrics = ['accuracy'])
```

```python
model.fit(xtrain, ytrain,
         epochs = 10,
         validation_data = (xtest,ytest))
```

```python
model.evaluate(xtrain,ytrain)
```

```python
model.evaluate(xtest,ytest)
```

```python
ytrain_pred = []

for pred in model.predict(xtrain):
    ytrain_pred.append(np.round(pred))


ytest_pred = []
for pred in model.predict(xtest):
    ytest_pred.append(np.round(pred))
```

```python
from sklearn.metrics import classification_report

print("Training classification report-")
print(classification_report(ytrain , ytrain_pred))
```

```python
print("Training classification report-")
print(classification_report(ytest , ytest_pred))
```
